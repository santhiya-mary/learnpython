```python
#Use Case 2: Sum of Even and Odd Numbers
n = int(input("Enter a positive integer: "))

even_sum = 0
odd_sum = 0

for number in range(1, n + 1):
    if number % 2 == 0:
        even_sum += number
    else:
        odd_sum += number

print("Sum of even numbers:", even_sum)
print("Sum of odd numbers:", odd_sum)
```
